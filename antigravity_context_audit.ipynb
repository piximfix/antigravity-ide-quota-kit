```python
import os
from collections import defaultdict

HEAVY_DIRS = {
    'node_modules', '.next', 'dist', 'build', 'out', '.turbo',
    'vendor', '.venv', 'venv', '__pycache__', '.git', '.pytest_cache'
}

def audit_workspace(root='.'):
    folder_bytes = defaultdict(int)
    for dirpath, dirnames, filenames in os.walk(root):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            try:
                size = os.path.getsize(fp)
            except OSError:
                continue
            parts = os.path.relpath(dirpath, root).split(os.sep)
            top = parts[0] if parts and parts[0] != '.' else '(root)'
            folder_bytes[top] += size
    for name, total in sorted(folder_bytes.items(), key=lambda x: -x[1])[:12]:
        tag = '[IGNORE CANDIDATE]' if name in HEAVY_DIRS else '[SOURCE]'
        print(f'{tag:20s} {total/1024:9.1f} KB  {name}')

audit_workspace('.')
```
